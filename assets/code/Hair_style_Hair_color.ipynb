```python
import os

def tree(root, max_depth=2, max_items=10):
    """
    Print a tree-like view of folders and files with constraints:
      - max_depth: how deep to traverse
      - max_items: maximum files/folders to show per directory (EXCEPT root)
    """
    def walk(dir_path, depth=0, prefix=""):
        if depth > max_depth:
            return

        try:
            items = sorted(os.listdir(dir_path))
        except PermissionError:
            return

        # Apply max_items only after root
        if depth > 0:
            items = items[:max_items]

        for i, name in enumerate(items):
            if name == '.git':
                continue

            path = os.path.join(dir_path, name)
            connector = "└── " if i == len(items) - 1 else "├── "
            print(prefix + connector + name)

            if os.path.isdir(path) and depth < max_depth:
                walk(
                    path,
                    depth + 1,
                    prefix + ("    " if i == len(items) - 1 else "│   ")
                )

    print(root)
    walk(root)
```

```python
import os
import math
from PIL import Image
import matplotlib.pyplot as plt

def plot_images_from_folder(folder_path, images_per_row=5, max_images=None):
    """
    Plot images from a folder in a grid with filenames above each image.

    Args:
        folder_path (str): Path to image folder
        images_per_row (int): Number of images per row (default=5)
        max_images (int or None): Limit number of images plotted
    """
    from PIL import Image
    image_files = sorted([
        f for f in os.listdir(folder_path)
        if f.lower().endswith(('.png', '.jpg', '.jpeg', '.webp'))
    ])

    if max_images:
        image_files = image_files[:max_images]

    if len(image_files) == 0:
        print("No images found in folder.")
        return

    rows = math.ceil(len(image_files) / images_per_row)
    plt.figure(figsize=(images_per_row * 3, rows * 3))

    for idx, img_name in enumerate(image_files):
        img_path = os.path.join(folder_path, img_name)
        img = Image.open(img_path).convert("RGB")

        ax = plt.subplot(rows, images_per_row, idx + 1)
        ax.imshow(img)
        ax.axis("off")
        ax.set_title(img_name, fontsize=8, pad=2)  # filename above image, small font

    plt.tight_layout()
    plt.show()
```

```python
%%writefile /kaggle/working/HairFastGAN/app.py


#!/usr/bin/env python3
# coding: utf-8

import os
import sys
import gradio as gr
import torch
from pathlib import Path
from PIL import Image
import numpy as np
from torchvision.utils import save_image
import torchvision.transforms.functional as F

# Add the current directory to path
sys.path.append(os.path.dirname(os.path.abspath(__file__)))

from hair_swap import HairFast, get_parser

# Initialize HairFast model
print("Loading HairFastGAN model...")
model_args = get_parser().parse_args([])
hair_fast = HairFast(model_args)
print("Model loaded successfully!")

# Example images directory
example_dir = "input"
color_ference_examples = "input/color_ference_examples"
face_image_examples = "input/face_image_examples"
shapere_ference_examples = "input/shapere_ference_examples"

# CSS for clean look and uniform sizing (matching LivePortrait style)
css = '''
#col-container {max-width: 1400px; margin: 0 auto;}
.tab-nav button {font-size: 16px; padding: 10px 20px;}
/* Make all image components the same size */
.image-container {
    height: 500px !important;
    width: 100% !important;
}
.image-container img {
    height: 500px !important;
    width: 100% !important;
    object-fit: contain !important;
}
'''

def convert_to_pil(image):
    """Convert various image formats to PIL Image"""
    if isinstance(image, str):
        return Image.open(image)
    elif isinstance(image, np.ndarray):
        return Image.fromarray(image)
    elif isinstance(image, Image.Image):
        return image
    return None

def hair_swap_interface(face_img, shape_img, color_img, mixing_value, smooth_value):
    """
    Main function for hair swapping
    """
    try:
        if face_img is None:
            return None, "❌ Please upload a face image"
        if shape_img is None:
            return None, "❌ Please upload a hair shape reference image"
        if color_img is None:
            return None, "❌ Please upload a hair color reference image"
        
        # Update model args with slider values
        model_args.mixing = mixing_value
        model_args.smooth = int(smooth_value)
        
        # Convert inputs to PIL if needed
        face_pil = convert_to_pil(face_img)
        shape_pil = convert_to_pil(shape_img)
        color_pil = convert_to_pil(color_img)
        
        # Run hair swap
        result_tensor = hair_fast(face_pil, shape_pil, color_pil, benchmark=False)
        
        # Convert result tensor to numpy array for Gradio
        result_np = result_tensor.squeeze(0).permute(1, 2, 0).cpu().numpy()
        result_np = (result_np * 255).astype(np.uint8)
        
        return result_np, "✅ Hair swap completed successfully!"
        
    except Exception as e:
        return None, f"❌ Error: {str(e)}"

def hair_swap_same_reference(face_img, reference_img, mixing_value, smooth_value):
    """
    Hair swap using the same image for both shape and color
    """
    return hair_swap_interface(face_img, reference_img, reference_img, mixing_value, smooth_value)

def create_side_by_side(face_img, shape_img, color_img, result_img):
    """
    Create a side-by-side comparison image
    """
    try:
        if result_img is None:
            return None
        
        # Convert all to PIL
        face_pil = convert_to_pil(face_img)
        shape_pil = convert_to_pil(shape_img)
        color_pil = convert_to_pil(color_img)
        result_pil = convert_to_pil(result_img)
        
        # Resize all to same size
        size = (256, 256)
        face_pil = face_pil.resize(size)
        shape_pil = shape_pil.resize(size)
        color_pil = color_pil.resize(size)
        result_pil = result_pil.resize(size)
        
        # Create concatenated image
        total_width = size[0] * 4
        combined = Image.new('RGB', (total_width, size[1]))
        combined.paste(face_pil, (0, 0))
        combined.paste(shape_pil, (size[0], 0))
        combined.paste(color_pil, (size[0] * 2, 0))
        combined.paste(result_pil, (size[0] * 3, 0))
        
        return np.array(combined)
    except:
        return None

# Build Gradio Interface
with gr.Blocks(theme=gr.themes.Soft(font=[gr.themes.GoogleFont("Plus Jakarta Sans")]), css=css) as demo:
    
    gr.Markdown("# 💇 HairFastGAN - Realistic Hair Transfer")
    gr.Markdown("Transfer hairstyle and color from reference images to your portrait in seconds")
    
    with gr.Tabs() as main_tabs:
        
        # ==================== TAB 1: FULL HAIR TRANSFER ====================
        with gr.TabItem("💇‍♀️ Full Hair Transfer"):
            gr.Markdown("### Transfer both hair shape and color from different reference images")
            
            with gr.Row():
                with gr.Column():
                    gr.Markdown("**👤 Face Image**")
                    face_image_full = gr.Image(type="filepath", label="Upload Face Image", height=500)
                    
                    if os.path.exists(face_image_examples):
                        example_files = [f for f in os.listdir(face_image_examples) if f.endswith(('.png', '.jpg', '.jpeg'))]
                        if example_files:
                            gr.Examples(
                                examples=[[os.path.join(face_image_examples, f)] for f in example_files[:4]],
                                inputs=[face_image_full],
                                label="Example Face Images"
                            )
                
                with gr.Column():
                    gr.Markdown("**✂️ Hair Shape Reference**")
                    shape_image_full = gr.Image(type="filepath", label="Upload Shape Reference", height=500)
                    
                    if os.path.exists(shapere_ference_examples):
                        example_files = [f for f in os.listdir(shapere_ference_examples) if f.endswith(('.png', '.jpg', '.jpeg'))]
                        if example_files:
                            gr.Examples(
                                examples=[[os.path.join(shapere_ference_examples, f)] for f in example_files[:4]],
                                inputs=[shape_image_full],
                                label="Example Shape References"
                            )
                
                with gr.Column():
                    gr.Markdown("**🎨 Hair Color Reference**")
                    color_image_full = gr.Image(type="filepath", label="Upload Color Reference", height=500)
                    
                    if os.path.exists(color_ference_examples):
                        example_files = [f for f in os.listdir(color_ference_examples) if f.endswith(('.png', '.jpg', '.jpeg'))]
                        if example_files:
                            gr.Examples(
                                examples=[[os.path.join(color_ference_examples, f)] for f in example_files[:4]],
                                inputs=[color_image_full],
                                label="Example Color References"
                            )
            
        
            
            with gr.Row():
                swap_btn_full = gr.Button("🚀 Transfer Hair", variant="primary", size="lg")
                clear_btn_full = gr.ClearButton(
                    [face_image_full, shape_image_full, color_image_full],
                    value="🧹 Clear All"
                )
            
            with gr.Row():
                with gr.Column():
                    output_image_full = gr.Image(type="numpy", label="✨ Result", height=500)
                with gr.Column():
                    comparison_full = gr.Image(type="numpy", label="📊 Side-by-Side Comparison", height=500)
            
            status_full = gr.Textbox(label="Status", interactive=False)
        
        # ==================== TAB 2: QUICK TRANSFER (SAME REFERENCE) ====================
        with gr.TabItem("⚡ Quick Transfer"):
            gr.Markdown("### Transfer hair using a single reference image (for both shape and color)")
            
            with gr.Row():
                with gr.Column():
                    gr.Markdown("**👤 Face Image**")
                    face_image_quick = gr.Image(type="filepath", label="Upload Face Image", height=500)
                    
                    if os.path.exists(example_dir):
                        example_files = [f for f in os.listdir(example_dir) if f.endswith(('.png', '.jpg', '.jpeg'))]
                        if example_files:
                            gr.Examples(
                                examples=[[os.path.join(example_dir, f)] for f in example_files[:4]],
                                inputs=[face_image_quick],
                                label="Example Face Images"
                            )
                
                with gr.Column():
                    gr.Markdown("**💇 Hair Reference (Shape + Color)**")
                    reference_image_quick = gr.Image(type="filepath", label="Upload Reference Image", height=500)
                    
                    if os.path.exists(example_dir):
                        example_files = [f for f in os.listdir(example_dir) if f.endswith(('.png', '.jpg', '.jpeg'))]
                        if example_files:
                            gr.Examples(
                                examples=[[os.path.join(example_dir, f)] for f in example_files[:4]],
                                inputs=[reference_image_quick],
                                label="Example References"
                            )
                
                with gr.Column():
                    output_image_quick = gr.Image(type="numpy", label="✨ Result", height=500)
            
            
            with gr.Row():
                swap_btn_quick = gr.Button("🚀 Transfer Hair", variant="primary", size="lg")
                clear_btn_quick = gr.ClearButton(
                    [face_image_quick, reference_image_quick],
                    value="🧹 Clear All"
                )
            
            status_quick = gr.Textbox(label="Status", interactive=False)
        
    # ==================== EVENT HANDLERS ====================
    
    # Full Transfer Tab
    def process_full_transfer(face, shape, color):
        DEFAULT_MIXING = 0.95
        DEFAULT_SMOOTH = 5
        result, status = hair_swap_interface(face, shape, color, DEFAULT_MIXING, DEFAULT_SMOOTH)
        comparison = create_side_by_side(face, shape, color, result) if result is not None else None
        return result, comparison, status

    swap_btn_full.click(
        fn=process_full_transfer,
        inputs=[face_image_full, shape_image_full, color_image_full],
        outputs=[output_image_full, comparison_full, status_full],
        show_progress=True
    )

    
    # Quick Transfer Tab
    def process_quick_transfer(face, reference):
        DEFAULT_MIXING = 0.95
        DEFAULT_SMOOTH = 5
        result, status = hair_swap_same_reference(face, reference, DEFAULT_MIXING, DEFAULT_SMOOTH)
        return result, status

    
    swap_btn_quick.click(
        fn=process_quick_transfer,
        inputs=[face_image_quick, reference_image_quick],
        outputs=[output_image_quick, status_quick],
        show_progress=True
    )

    

# Launch the app
if __name__ == "__main__":
    import argparse
    
    parser = argparse.ArgumentParser()
    parser.add_argument("--share", action="store_true", help="Create a public link")
    parser.add_argument("--server_port", type=int, default=7860, help="Server port")
    parser.add_argument("--server_name", type=str, default="0.0.0.0", help="Server name")
    args = parser.parse_args()
    
    demo.launch(
        server_port=args.server_port,
        share=args.share,
        server_name=args.server_name
    )
    
```
